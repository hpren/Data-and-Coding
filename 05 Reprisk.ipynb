Reprisk 2019

```python
reader = pd.read_stata(
    r"D:\Dropbox\Dropbox\Dropbox\4_SASB\Shared Data\RepRisk\Metrics_2019_Reprisk.dta\estpgmvpwahjtppf.dta",
    iterator=True,
    chunksize=100000
)

us_chunks = []

for chunk in reader:
# import data in chunks and filter for US companies with a primary ISIN
    us = chunk[
        (chunk['headquarters_country_isocode'].str.strip() == 'US') &
        (chunk['primary_isin'].str.strip() != '')
    ]

    if len(us) > 0:
        us_chunks.append(us)

reprisk2019 = pd.concat(us_chunks, ignore_index=True)

print(len(reprisk2019))
```

```python
reprisk2019 = reprisk2019[
    reprisk2019['primary_isin'].str.strip() != ''
].copy()

reprisk2019 = reprisk2019[['reprisk_id', 'company_name','date', 'current_rri', 'trend_rri', 'peak_rri', 'reprisk_rating']]
reprisk2019.head()
```
